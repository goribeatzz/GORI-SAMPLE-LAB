```python
# GORI SAMPLE LAB — ACE-Step 1.5 gratuito no Google Colab
# Use GPU T4: Ambiente de execução > Alterar tipo de ambiente de execução > T4 GPU.
# Esta célula instala Python 3.12 isolado, instala o projeto oficial e abre a interface web.
# Não requer Hugging Face token nem repositório gated.

import os, subprocess, sys, pathlib, textwrap

ROOT = "/content/ACE-Step-1.5"

def run(cmd, cwd=None, env=None):
    print("\n>>>", " ".join(cmd) if isinstance(cmd, list) else cmd, flush=True)
    subprocess.check_call(cmd, cwd=cwd, env=env)

print("1/4 — Verificando GPU do Colab...")
try:
    run(["nvidia-smi"])
except Exception:
    print("ATENÇÃO: nenhuma GPU NVIDIA foi detectada. Configure T4 GPU antes de continuar.")

print("\n2/4 — Instalando uv e Python 3.12 isolado...")
run([sys.executable, "-m", "pip", "install", "-q", "uv"])
run(["uv", "python", "install", "3.12"])

print("\n3/4 — Baixando e instalando ACE-Step 1.5 oficial...")
if not os.path.isdir(ROOT):
    run(["git", "clone", "--depth", "1", "https://github.com/ACE-Step/ACE-Step-1.5.git", ROOT])
else:
    print("Repositório já existe; reutilizando arquivos.")
run(["uv", "sync", "--python", "3.12"], cwd=ROOT)

print("\n4/4 — Abrindo ACE-Step 1.5...")
env = os.environ.copy()
env["ACESTEP_LM_MODEL_PATH"] = "acestep-5Hz-lm-0.6B"
env["ACESTEP_INIT_LLM"] = "auto"
env["PYTHONUNBUFFERED"] = "1"

print("\nAguarde: na primeira execução, os modelos serão baixados (vários GB).")
print("Quando aparecer um link público gradio.live, abra-o para usar o gerador.")
print("Mantenha esta célula em execução enquanto estiver gerando.")
run([
    "uv", "run", "--python", "3.12", "acestep",
    "--server-name", "0.0.0.0",
    "--port", "7860",
    "--share",
    "--language", "en",
    "--config_path", "acestep-v15-turbo",
    "--lm_model_path", "acestep-5Hz-lm-0.6B",
    "--init_llm", "auto",
    "--offload_to_cpu"
], cwd=ROOT, env=env)
```
